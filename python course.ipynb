#### 2. Bitwise operators(<<, >>, ~, &, |,^) 

```python
#bitwise operators works on binary number
#bitwise operators are faster than arithmatic operators
#bitwise operators perform many operation in easy way due to binary number system
```

```python
#ADD Operator (&) If both the bit are true answer is true
#OR Operator (|) If one of the bit is true answer is true
#XOR Operator (^) If both bit are differnt then answer is True (if we have even number of 1's output is 0, if we have odd number of 1 = 1)
```

```python
a = 10
b = 3

print(a & b)
print(a | b)
print(a ^ b)
```

left shift(<<), Right shift(>>)

```python
# Def (left shift): Insert new bits (0) to the LSB (number is doubing)
# Def (Right shift): Delete the bits from the LSB (number is half)
```

```python
a = 10
print(a<< 1)
print(a << 2)
print(a << 3) #it works as (a * (2  power k))
print(a << 6) #(10 * (2**6))
```

```python
a = 10
print(a >> 1) #it works as (a // (2 power K))
print(a >> 2) # (10 // (2**2)
print(a >> 3)
print(a >> 4)
```

bitwise not (~)

```python
#def (not) : flip the bit 
a = 12 #(if we take postive number output will be negative + 1)
print(~a)
```

```python
a = -12 #(if we take negative number output will be positive - 1)
print(~a)
```

# match case

```python
# Match case is used to replace multiple if else condition
# Match case is more powerful than switch case because match case works with variables & data Structurs
```

```python
# Example 1
num = int(input())

match num:
    case 10:
        print("you entered 10")
    case 20:
        print("you entered 20")
    case 30:
        print ("you entered 30")
    case _:
        print("something else")
```

```python
# Example 2 : passind data structure to match case

data = [1,2,3,4]
match data:
    case [4,3,2,1]:
        print("Hello")
    case [1,2,3,4]:
        print("world")
```

```python
# example
Biryani = input("Famous Hyderabadi biryani")
match Biryani:
    case "Paradise":
        print("Spicy & tasty")
    case "Shahdab":
        print("double masala")
    case "road side":
        print("budget friendly")
    case _:
        print("something else")
        
```

```python
#Example 3: if condition with match case

val = int(input())

match val:
    case x if x > 0:
        print("it is positive number")
    case y if y < 0:
        print("it is negative number")
    case _:
        print("it is a zero")
```

```python
# we have negative and positive indexing
# string[start:end:step]  start = 0, end = n-1, step = 1

Heroine = "Samantha"

print(Heroine[4])
print(Heroine[-5])

print(Heroine[0:5])
print(Heroine[0:5:3])
print(Heroine[::])
print(Heroine[::2])
print(Heroine[6:3:1])   # if step size is positive we move to the right side & we r 

print(Heroine[6:3:-1])  # for -ve indexing we move left side and end = end - (-1) = end + 1
print(Heroine[-1 : -6])  # by default step size is 1 as +ve step we move right size & it print empty
print(Heroine[-1:-6:-1])
print(Heroine[-7::])     # by default step size is 1 as +ve step we move right size 
print(Heroine[-7::])

print(Heroine[0:-1:])    # combined +ve And -ve

print(Heroine[-10::])   # when we start at out of range we start from the last in ve
print(Heroine[-1:-6:-2])
print(Heroine[10:-6:])
print(Heroine[0:2:1] + Heroine[5::1])
```

```python
# tuple item cannot be updated directly
# complete item in tuple cannot be updated directly
# but if item is mutable then elementsinside this mutable item can be updated
```

```python
# insert
tuple3 = (10,20,30)
list1 = list(tuple3)
print(list1)
list1.extend([100,200,300])
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
#delete
tuple3 = (10,20,30)
list1 = list(tuple3)
print(list1)
list1.remove(30)
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
#update
tuple3 = (10,20,30)
list1 = list(tuple3)
print(list1)
list1[2] = 3000
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

# popular inbuilt function for tuple

```python
salary = (6,70,100,600,700,90,10,80,120)

print(tuple(reversed(salary)))
print(len(salary))
print(min(salary))
print(max(salary))
print(sum(salary))

sorted_list = tuple(sorted(salary, reverse = True))
print(sorted_list)

# sort() is inplace function
# inplace means sorting is done in the same data structure

# sorted() function return the list
# sort() function don't return anything
```
